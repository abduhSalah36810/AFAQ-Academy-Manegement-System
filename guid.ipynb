```python
import sqlite3


DATABASE_NAME = "dafaq.db"


def get_connection():
    return sqlite3.connect(DATABASE_NAME)


def create_tables():
    connection = get_connection()

    connection.execute("""
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            email TEXT UNIQUE NOT NULL,
            role TEXT NOT NULL
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS courses (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            instructor_id INTEGER
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS enrollments (
            trainee_id INTEGER,
            course_id INTEGER,
            PRIMARY KEY (trainee_id, course_id)
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS attendance (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            trainee_id INTEGER,
            course_id INTEGER,
            date TEXT NOT NULL,
            status TEXT NOT NULL
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS grades (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            trainee_id INTEGER,
            course_id INTEGER,
            score REAL NOT NULL
        )
    """)

    connection.commit()
    connection.close()


if __name__ == "__main__":
    create_tables()
    print("Database initialized successfully.")
```

```python
class TraineeService:

    def __init__(self, connection):
        self.connection = connection

    def create(self, trainee):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            INSERT INTO users (name, email, role)
            VALUES (?, ?, ?)
            """,
            (trainee.name, trainee.email, "trainee")
        )

        self.connection.commit()

    def get_all(self):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            SELECT id, name, email
            FROM users
            WHERE role = 'trainee'
            """
        )

        return cursor.fetchall()
```

```python
class EnrollmentService:

    def __init__(self, connection):
        self.connection = connection

    def enroll(self, trainee_id, course_id):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            INSERT INTO enrollments (trainee_id, course_id)
            VALUES (?, ?)
            """,
            (trainee_id, course_id)
        )

        self.connection.commit()

    def get_trainee_courses(self, trainee_id):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            SELECT courses.id, courses.name
            FROM courses
            JOIN enrollments
                ON courses.id = enrollments.course_id
            WHERE enrollments.trainee_id = ?
            """,
            (trainee_id,)
        )

        return cursor.fetchall()
```

ثم تقول:

"هنا ظهر عندنا SQL أهم شوية:

`JOIN`.

```python
from database import get_connection, create_tables

from models.trainee import Trainee
from models.course import Course

from services.trainee_service import TraineeService
from services.course_service import CourseService
from services.enrollment_service import EnrollmentService


def main():

    create_tables()

    connection = get_connection()

    trainee_service = TraineeService(connection)
    course_service = CourseService(connection)
    enrollment_service = EnrollmentService(connection)

    while True:

        print("\n==============================")
        print("       DAFAQ ACADEMY")
        print("==============================")
        print("1. Add Trainee")
        print("2. View Trainees")
        print("3. Add Course")
        print("4. View Courses")
        print("5. Enroll Trainee")
        print("6. View Trainee Courses")
        print("7. Exit")

        choice = input("\nChoose an option: ")

        if choice == "1":

            name = input("Trainee name: ")
            email = input("Trainee email: ")

            trainee = Trainee(name, email)

            trainee_service.create(trainee)

            print("Trainee added successfully.")

        elif choice == "2":

            trainees = trainee_service.get_all()

            print("\n--- Trainees ---")

            for trainee in trainees:
                print(
                    f"ID: {trainee[0]} | "
                    f"Name: {trainee[1]} | "
                    f"Email: {trainee[2]}"
                )

        elif choice == "3":

            name = input("Course name: ")

            course = Course(name)

            course_service.create(course)

            print("Course created successfully.")

        elif choice == "4":

            courses = course_service.get_all()

            print("\n--- Courses ---")

            for course in courses:
                print(
                    f"ID: {course[0]} | "
                    f"Name: {course[1]}"
                )

        elif choice == "5":

            trainee_id = int(input("Trainee ID: "))
            course_id = int(input("Course ID: "))

            enrollment_service.enroll(
                trainee_id,
                course_id
            )

            print("Trainee enrolled successfully.")

        elif choice == "6":

            trainee_id = int(input("Trainee ID: "))

            courses = enrollment_service.get_trainee_courses(
                trainee_id
            )

            print("\n--- Enrolled Courses ---")

            for course in courses:
                print(
                    f"ID: {course[0]} | "
```

```python
import sqlite3


DATABASE_NAME = "dafaq.db"


def get_connection():
    connection = sqlite3.connect(DATABASE_NAME)

    # Enable foreign key enforcement in SQLite.
    connection.execute("PRAGMA foreign_keys = ON")

    return connection


def create_tables():
    connection = get_connection()

    connection.execute("""
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            email TEXT UNIQUE NOT NULL,
            password_hash TEXT NOT NULL,
            role TEXT NOT NULL
                CHECK (role IN ('trainee', 'instructor', 'admin'))
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS courses (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            instructor_id INTEGER,

            FOREIGN KEY (instructor_id)
                REFERENCES users(id)
                ON DELETE SET NULL
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS enrollments (
            trainee_id INTEGER NOT NULL,
            course_id INTEGER NOT NULL,

            PRIMARY KEY (trainee_id, course_id),

            FOREIGN KEY (trainee_id)
                REFERENCES users(id)
                ON DELETE CASCADE,

            FOREIGN KEY (course_id)
                REFERENCES courses(id)
                ON DELETE CASCADE
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS attendance (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            trainee_id INTEGER NOT NULL,
            course_id INTEGER NOT NULL,
            date TEXT NOT NULL,
            status TEXT NOT NULL
                CHECK (status IN ('present', 'absent')),

            UNIQUE (trainee_id, course_id, date),

            FOREIGN KEY (trainee_id)
                REFERENCES users(id)
                ON DELETE CASCADE,

            FOREIGN KEY (course_id)
                REFERENCES courses(id)
                ON DELETE CASCADE
        )
    """)

    connection.execute("""
        CREATE TABLE IF NOT EXISTS grades (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            trainee_id INTEGER NOT NULL,
            course_id INTEGER NOT NULL,
            score REAL NOT NULL
                CHECK (score >= 0 AND score <= 100),

            UNIQUE (trainee_id, course_id),

            FOREIGN KEY (trainee_id)
                REFERENCES users(id)
                ON DELETE CASCADE,

            FOREIGN KEY (course_id)
                REFERENCES courses(id)
                ON DELETE CASCADE
        )
    """)

    # Indexes for frequently searched columns.
    connection.execute("""
        CREATE INDEX IF NOT EXISTS idx_users_email
        ON users(email)
    """)

    connection.execute("""
        CREATE INDEX IF NOT EXISTS idx_enrollments_trainee
        ON enrollments(trainee_id)
    """)

    connection.execute("""
        CREATE INDEX IF NOT EXISTS idx_enrollments_course
        ON enrollments(course_id)
    """)

    connection.execute("""
        CREATE INDEX IF NOT EXISTS idx_attendance_trainee
        ON attendance(trainee_id)
    """)

    connection.execute("""
        CREATE INDEX IF NOT EXISTS idx_grades_trainee
        ON grades(trainee_id)
    """)

    connection.commit()
    connection.close()


if __name__ == "__main__":
    create_tables()
    print("Database initialized successfully.")
```

```python
import hashlib
import hmac
import secrets


ITERATIONS = 300_000


def hash_password(password):
    salt = secrets.token_bytes(16)

    password_hash = hashlib.pbkdf2_hmac(
        "sha256",
        password.encode("utf-8"),
        salt,
        ITERATIONS
    )

    return (
        salt.hex()
        + ":"
        + password_hash.hex()
    )


def verify_password(password, stored_hash):
    try:
        salt_hex, hash_hex = stored_hash.split(":")

        salt = bytes.fromhex(salt_hex)
        expected_hash = bytes.fromhex(hash_hex)

        actual_hash = hashlib.pbkdf2_hmac(
            "sha256",
            password.encode("utf-8"),
            salt,
            ITERATIONS
        )

        return hmac.compare_digest(
            actual_hash,
            expected_hash
        )

    except (ValueError, TypeError):
        return False
```

```python
from utils.security import hash_password, verify_password
from utils.validation import (
    normalize_email,
    is_valid_email,
    is_valid_password,
    validate_name
)


class AuthService:

    def __init__(self, connection):
        self.connection = connection

    def signup(self, name, email, password):
        name = name.strip()
        email = normalize_email(email)

        if not validate_name(name):
            return False, "Name must contain at least 2 characters."

        if not is_valid_email(email):
            return False, "Invalid email."

        if not is_valid_password(password):
            return False, "Password must be at least 8 characters."

        password_hash = hash_password(password)

        try:
            cursor = self.connection.cursor()

            cursor.execute(
                """
                INSERT INTO users
                    (name, email, password_hash, role)
                VALUES
                    (?, ?, ?, ?)
                """,
                (
                    name,
                    email,
                    password_hash,
                    "trainee"
                )
            )

            self.connection.commit()

            return True, "Account created successfully."

        except Exception:
            self.connection.rollback()
            return False, "Email already exists or signup failed."

    def login(self, email, password):
        email = normalize_email(email)

        cursor = self.connection.cursor()

        cursor.execute(
            """
            SELECT id, name, email, password_hash, role
            FROM users
            WHERE email = ?
            """,
            (email,)
        )

        user = cursor.fetchone()

        if user is None:
            return None, "Invalid email or password."

        user_id, name, email, stored_hash, role = user

        if not verify_password(password, stored_hash):
            return None, "Invalid email or password."

        return {
            "id": user_id,
            "name": name,
            "email": email,
            "role": role
        }, "Login successful."
```

```python
    def add_grade(self, trainee_id, course_id, score):
        if score < 0 or score > 100:
            raise ValueError(

        cursor = self.connection.cursor()

            """
            INSERT INTO grades
                (trainee_id, course_id, score)
```

````python
import sqlite3

from utils.security import hash_password
from utils.validation import (
    normalize_email,
    is_valid_email,
    is_valid_password,
    validate_name
)


class AdminService:

    def __init__(self, connection):
        self.connection = connection

    def create_user(self, name, email, password, role):
        name = name.strip()
        email = normalize_email(email)

        allowed_roles = {
            "trainee",
            "instructor",
            "admin"
        }

        if role not in allowed_roles:
            raise ValueError("Invalid role.")

        if not validate_name(name):
            raise ValueError(
                "Name must contain at least 2 characters."
            )

        if not is_valid_email(email):
            raise ValueError("Invalid email.")

        if not is_valid_password(password):
            raise ValueError(
                "Password must be at least 8 characters."
            )

        password_hash = hash_password(password)

        try:
            cursor = self.connection.cursor()

            cursor.execute(
                """
                INSERT INTO users
                    (name, email, password_hash, role)
                VALUES
                    (?, ?, ?, ?)
                """,
                (
                    name,
                    email,
                    password_hash,
                    role
                )
            )

            self.connection.commit()

        except sqlite3.IntegrityError:
            self.connection.rollback()
            raise ValueError(
                "This email already exists."
            )

    def get_instructors(self):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            SELECT id, name, email
            FROM users
            WHERE role = 'instructor'
            ORDER BY name
            """
        )

        return cursor.fetchall()

    def get_trainees(self):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            SELECT id, name, email
            FROM users
            WHERE role = 'trainee'
            ORDER BY name
            """
        )

        return cursor.fetchall()

    def assign_instructor(self, instructor_id, course_id):
        cursor = self.connection.cursor()

        cursor.execute(
            """
            SELECT id
            FROM users
            WHERE id = ?
            AND role = 'instructor'
            """,
            (instructor_id,)
        )

        instructor = cursor.fetchone()

        if instructor is None:
            raise ValueError(
                "Instructor does not exist."
            )

        cursor.execute(
            """
            SELECT id
            FROM courses
            WHERE id = ?
            """,
            (course_id,)
        )

        course = cursor.fetchone()

        if course is None:
            raise ValueError(
                "Course does not exist."
            )

        cursor.execute(
            """
            UPDATE courses
            SET instructor_id = ?
            WHERE id = ?
            """,
            (
                instructor_id,
                course_id
            )
        )

        self.connection.commit()

لاحظ هنا حاجة مهمة جدًا:

الـ AdminService بيتعامل داخليًا مع:

```python
instructor_id
course_id
````
